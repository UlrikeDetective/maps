## Geofacets vs. Choropleth Maps: Fixing Spatial Bias in Python

Why traditional maps fail at regional data visualization — and how grid-based geofacets fix the problem.

If you’ve ever looked at a map of the United States shaded by data — a choropleth map — you’ve likely been lied to by geography.

Imagine we are mapping Electric Vehicle (EV) adoption rates by state. Take a look at the traditional choropleth map below:

On this standard map, large states like Montana and Wyoming — which have very low EV adoption (light gray) — dominate the screen with sweeping blocks of empty color. Meanwhile, states with the highest adoption rates (deep teal), like Massachusetts, Rhode Island, and New Jersey, are virtually invisible because of their tiny physical size.

This is known as Spatial Visual Bias — which occurs when geographic regions with large physical landmasses dominate a visualization, artificially skewing the reader’s perception. Physical landmass rarely correlates with the data we actually care about, such as population density, sales volume, or tech adoption. When massive states dominate the map, they distort the story, making it look like nobody is buying EVs when, in fact, millions of people in small states are.

So, how do we fix this? Enter the geofacet.

#### Curing Spatial Bias with Geofacets
A geofacet is a data visualization technique that abstracts geographic regions into a uniform grid of equal-sized cells, eliminating spatial bias while preserving regional awareness.

Instead of drawing the exact jagged outline of every state, geofacets assign each state a perfectly equal square. These squares are arranged to approximate the physical geography — Maine stays in the top right, Florida in the bottom right, and Texas anchors the bottom center.

To see the difference, let’s plot the exact same EV adoption data using a geofacet heatmap. Using the open-source Python package clean-charts, we can generate this with just a few lines of code:

```python
import pandas as pd
from clean_charts.plots import plot_geofacet
%matplotlib inline
import matplotlib.pyplot as plt

# Fictionalized EV adoption data for demonstration
data = {
    "AL": 12, "AK": 14, "AZ": 35, "AR": 11, "CA": 85, "CO": 50, "CT": 65, "DE": 55, "FL": 40, "GA": 35,
    "HI": 60, "ID": 18, "IL": 45, "IN": 25, "IA": 20, "KS": 18, "KY": 15, "LA": 14, "ME": 40, "MD": 65,
    "MA": 75, "MI": 30, "MN": 35, "MS": 10, "MO": 20, "MT": 15, "NE": 18, "NV": 55, "NH": 45, "NJ": 70,
    "NM": 25, "NY": 75, "NC": 40, "ND": 12, "OH": 28, "OK": 15, "OR": 70, "PA": 45, "RI": 60, "SC": 25,
    "SD": 14, "TN": 22, "TX": 35, "UT": 35, "VT": 65, "VA": 55, "WA": 78, "WV": 12, "WI": 25, "WY": 10, "DC": 80
}

ev_data = pd.DataFrame({
    'State': list(data.keys()),
    'EV_Adoption': list(data.values())
})

# Generate a heatmap grid
plot_geofacet(
    data=ev_data, 
    display_type="text",
    start_color="#EAEAEA", 
    end_color="#00695C", 
    title="EV Adoption Hotspots",
    subtitle="West Coast and Northeast lead the charge",
    output_path="ev_heatmap.png"
)

plt.show()
```

When viewed side-by-side with the traditional map, the advantage is instantly clear. By giving every state the exact same square footage, the geographic clusters in the Northeast are finally visible. The deep teal cells of the West Coast and Northeast are given equal visual weight to the light gray cells of the Midwest. Spatial bias is completely eliminated, creating a fair and accurate visual playing field.

#### Beyond Just Color: The Clean Canvas
The best part of using geofacets isn’t just the fairness — it’s that they give you a clean, isolated canvas inside every single cell.

Traditional maps force you to rely almost entirely on color shading. If you want to show two metrics at once, you’re forced to use awkward overlapping bubbles or confusing bivariate color scales. Geofacets, on the other hand, let you embed actual mini-charts inside each state’s square. Let’s look at two more ways to visualize this data inside a geofacet grid.

#### 1. The Donut Chart
Heatmaps are great for spotting trends, but what if you want to emphasize the remaining gap in the market? A donut chart provides an immediate part-to-whole relationship. Here, the deep teal slice represents EV sales, and the light gray remainder represents traditional gas vehicles.

On a traditional map, trying to draw 51 pie charts is a nightmare — the charts in the Northeast overlap into an illegible mess. Geofacets give every state an isolated cell, letting you instantly see that California’s donut is mostly teal, while Wyoming’s is mostly gray.

```python
# Generate a grid of donuts
plot_geofacet(
    data=ev_data, 
    display_type="donut",
    start_color="#EAEAEA", 
    end_color="#00695C", 
    title="The Remaining Market Gap",
    subtitle="EV Sales (Teal) vs. Gas Sales (Gray)",
    output_path="ev_donut.png"
)

plt.show()
```

#### 2. The Micro-Bar
When you want to track linear progress from 0% to 100%, a micro-bar is incredibly effective. It turns each state into a miniature progress bar. You can easily scan the grid and compare the lengths of the teal bars to see exactly how far each state has progressed toward full electrification, without any spatial distortion distracting you from the actual data.

```python
# Generate a grid of micro-bars
plot_geofacet(
    data=ev_data, 
    display_type="bar",
    start_color="#EAEAEA", 
    end_color="#00695C", 
    title="Progress Toward Total Electrification",
    subtitle="0% to 100% EV Adoption by State",
    output_path="ev_bar.png"
)

plt.show()
```
